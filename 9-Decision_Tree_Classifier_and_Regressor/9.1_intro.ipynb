# Decision Tree Classifier

## 1. Introduction

A Decision Tree Classifier is a supervised machine learning algorithm used for classification and regression tasks.  
It works by splitting the dataset into subsets based on feature values, forming a tree-like structure.

Two popular Decision Tree algorithms:
- ID3 (uses Entropy and Information Gain)
- CART (uses Gini Index)

Main concepts:
1. Entropy and Gini Index → Measure purity of data
2. Information Gain → Select best feature for splitting

---

## 2. Concept of Purity in Decision Trees

When a node contains data from only one class, it is called a **pure node**.  
When a node contains data from multiple classes, it is called an **impure node**.

Purity is measured using:
- Entropy
- Gini Impurity

---

## 3. Entropy

Entropy measures the uncertainty or randomness in data.

Formula of Entropy:

H(S) = - (p1 * log2(p1) + p2 * log2(p2) + ... + pn * log2(pn))

Where:
- p1, p2, ..., pn are probabilities of different classes.

### Example (Binary Classification)

Suppose a dataset has:
- Positive samples = p+
- Negative samples = p-

Then entropy is:

H(S) = - (p+ * log2(p+) + p- * log2(p-))

#### Case 1: Impure Split

Example:
- 3 Yes, 3 No

p+ = 3/6  
p- = 3/6  

H(S) = - (3/6 * log2(3/6) + 3/6 * log2(3/6))

This gives high entropy → Impure split.

#### Case 2: Pure Split

Example:
- 3 Yes, 0 No

p+ = 3/3  
p- = 0/3  

H(S) = - (1 * log2(1) + 0 * log2(0)) = 0

Entropy = 0 → Pure split.

---

## 4. Gini Impurity

Gini Impurity measures how often a randomly chosen element would be incorrectly classified.

Formula of Gini Impurity:

Gini = 1 - Σ (pi)^2

Where:
- pi is the probability of class i.

### Example (Binary Classification)

If:
p1 = 1/2  
p2 = 1/2  

Then:

Gini = 1 - ((1/2)^2 + (1/2)^2)
Gini = 1 - (1/4 + 1/4)
Gini = 1 - 1/2
Gini = 0.5 → Impure split

If:
p1 = 1  
p2 = 0  

Then:

Gini = 1 - (1^2 + 0^2)
Gini = 0 → Pure split

---

## 5. Information Gain

Information Gain is used to select the best feature for splitting in a decision tree.

Formula:

Gain(S, F) = H(S) - Σ ( |Sv| / |S| * H(Sv) )

Where:
- S = total dataset
- F = feature
- Sv = subset after splitting on feature F
- |S| = total samples
- |Sv| = samples in subset
- H(S) = entropy of parent node
- H(Sv) = entropy of child nodes

---

## 6. Example of Information Gain Calculation

Suppose a dataset has 14 samples:
- 9 Yes
- 5 No

### Step 1: Entropy of Parent Node

p+ = 9/14  
p- = 5/14  

H(S) = - (9/14 * log2(9/14) + 5/14 * log2(5/14))

H(S) ≈ 0.94

---

### Step 2: Split by Feature F1

Suppose feature F1 splits data into two subsets:

Subset C1:
- 6 Yes
- 2 No  
Total = 8

Subset C2:
- 3 Yes
- 3 No  
Total = 6

#### Entropy of C1

p+ = 6/8  
p- = 2/8  

H(C1) = - (6/8 * log2(6/8) + 2/8 * log2(2/8))

H(C1) ≈ 0.81

#### Entropy of C2

p+ = 3/6  
p- = 3/6  

H(C2) = - (3/6 * log2(3/6) + 3/6 * log2(3/6))

H(C2) = 1

---

### Step 3: Weighted Entropy After Split

Weighted Entropy = (8/14 * 0.81) + (6/14 * 1)

Weighted Entropy ≈ 0.95

---

### Step 4: Information Gain

Gain(S, F1) = H(S) - Weighted Entropy  
Gain(S, F1) = 0.94 - 0.95  
Gain(S, F1) ≈ 0.049

---

### Step 5: Feature Selection

If another feature F2 gives:

Gain(S, F2) = 0.051

Since:

Gain(S, F2) > Gain(S, F1)

Therefore, feature F2 is chosen for splitting.

---

## 7. Entropy vs Gini Impurity

Entropy:
H(S) = - Σ (pi * log2(pi))

Gini Impurity:
Gini = 1 - Σ (pi)^2

Comparison:
- Entropy is more sensitive to changes in probability.
- Gini is faster to compute.
- ID3 uses Entropy.
- CART uses Gini Impurity.

When difference in probabilities is small → Entropy is preferred.  
When difference is large → Gini Impurity is preferred.

---

## 8. Conclusion

Decision Tree construction involves:
1. Measuring purity using Entropy or Gini.
2. Calculating Information Gain for each feature.
3. Selecting the feature with maximum Information Gain.
4. Repeating the process recursively until pure nodes are obtained.

This is how a Decision Tree Classifier is built.
