# ✂️ Pruning in Decision Trees

**Topic:** Pre-Pruning and Post-Pruning  
**Purpose:** To control overfitting and improve generalization of Decision Trees.


## 1. Why is Pruning Used in Decision Trees?

Decision Trees can grow very large and complex.

### 🔴 Problem: Overfitting
- The tree learns **noise and outliers** in training data.
- It performs very well on training data.
- But it performs **poorly on new (unseen) data**.

This situation is called **Overfitting**.

### ✅ Solution: Pruning
**Pruning** means cutting unnecessary branches of the tree to:

- Reduce model complexity
- Improve generalization
- Increase prediction accuracy on test data
- Make the tree simpler and easier to interpret


## 2. Pre-Pruning (Early Stopping)

**Definition:**  
Pre-pruning means **stopping the growth of the decision tree early**, before it becomes too deep.

The tree is prevented from splitting further if certain conditions are met.


## 3. How Pre-Pruning Works

The decision tree stops splitting a node when:

- Maximum depth of tree is reached  
- Information Gain is below a threshold  
- Number of samples at a node is too small  
- Gini / Entropy improvement is very small  

### ✨ Example:
If a node has very few samples (e.g., only 2 or 3 data points),  
the algorithm will **not split further** and will make it a leaf node.

👉 This prevents the tree from learning noise.


## 4. Pre-Pruning: Advantages and Disadvantages

### ✅ Advantages:
- Faster training (tree grows smaller)
- Reduces overfitting
- Simple and computationally efficient
- Controls tree size

### ❌ Disadvantages:
- May stop too early
- Can cause **underfitting**
- May miss important patterns in data


## 5. Post-Pruning

**Definition:**  
Post-pruning means **first growing the full decision tree**, and then **removing unnecessary branches** after the tree is built.

In short:  
👉 Grow full tree → Then prune back.


## 6. How Post-Pruning Works

Steps:

1. Grow the complete decision tree until all leaves are pure or no further split is possible.
2. Evaluate subtrees using:
   - Validation set
   - Cross-validation
3. Remove (prune) branches that do not improve performance on validation data.
4. Replace pruned subtrees with leaf nodes.

👉 Only those branches that **do not help generalization** are removed.


## 7. Post-Pruning: Advantages and Disadvantages

### ✅ Advantages:
- Usually gives **better accuracy** than pre-pruning
- Reduces overfitting effectively
- Keeps important structure of the tree
- More reliable model

### ❌ Disadvantages:
- More computation required
- Slower training
- Needs validation data or cross-validation


## 8. Pre-Pruning vs Post-Pruning

| Feature        | Pre-Pruning                     | Post-Pruning                    |
|---------------|----------------------------------|----------------------------------|
| When applied  | During tree growth               | After full tree is built         |
| Tree size     | Smaller from start               | Large first, then reduced        |
| Risk          | Underfitting                     | Less underfitting                |
| Accuracy      | May be lower                     | Usually higher                   |
| Computation   | Low                              | High                             |
| Flexibility   | Less flexible                    | More flexible                    |


## 9. Exam-Ready Conclusion

Pruning is used to prevent **overfitting in Decision Trees**.

- **Pre-pruning** controls growth early but may cause underfitting.
- **Post-pruning** simplifies a fully grown tree and usually gives better generalization.

Hence, pruning improves the **predictive performance and robustness** of Decision Tree models.
