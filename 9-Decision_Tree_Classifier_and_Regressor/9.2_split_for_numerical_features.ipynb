# 🌳 Decision Tree Split for Numerical Features

**Topic:** Splitting Numerical (Continuous) Attributes in Decision Trees  
**Example:** Play Tennis Dataset  
**Objective:** Learn how thresholds are chosen using Information Gain / Gini Index.

## 1. Numerical Feature (Continuous Attribute)

A **numerical feature** is a feature that takes continuous values.

**Examples:**
- Temperature (23°C, 36°C, 44°C)
- Humidity (%)
- Age
- Salary

👉 Decision Trees cannot directly split continuous values like categories.  
So, they convert numerical features into **binary splits using thresholds**.

**Example split:**
Temperature ≤ 30  vs  Temperature > 30


## 2. Example Dataset (Play Tennis)

**Target Variable:** Play Tennis (Yes / No)  
**Feature Considered:** Temperature (Numerical)

| Day | Temperature | Play Tennis |
|-----|-------------|-------------|
| 1   | 23          | Yes         |
| 2   | 36          | Yes         |
| 3   | 44          | No          |
| 4   | 52          | No          |
| 5   | 67          | Yes         |
| 6   | 89          | No          |
| 7   | 105         | Yes         |


## 3. Step 1: Sort the Numerical Feature

First, sort the data based on the numerical feature (Temperature):

Temperature:  
23, 36, 44, 52, 67, 89, 105  

Corresponding Labels:  
Yes, Yes, No, No, Yes, No, Yes  

👉 Sorting helps to find possible split points (thresholds).


## 4. Step 2: Find Possible Thresholds

Thresholds are taken as **midpoints between consecutive values**.

Example thresholds:

Between 23 and 36 → (23 + 36) / 2 = 29.5  
Between 36 and 44 → (36 + 44) / 2 = 40  
Between 44 and 52 → (44 + 52) / 2 = 48  
Between 52 and 67 → (52 + 67) / 2 = 59.5  

Each threshold creates a binary split:

Temperature ≤ T   vs   Temperature > T


## 5. Step 3: Create Binary Splits

For a chosen threshold T:

Left Node  → Temperature ≤ T  
Right Node → Temperature > T  

Example:

If T = 40  

Left: 23, 36  
Right: 44, 52, 67, 89, 105  

Each node contains its own Yes/No labels.


## 6. Step 4: Information Gain

To choose the **best threshold**, we compute **Information Gain**.

Information Gain measures how much **uncertainty (entropy) is reduced** after the split.

Formula:

Information Gain =  
Entropy(parent) − [  
   (|Left| / |Total|) × Entropy(Left)  
 + (|Right| / |Total|) × Entropy(Right)  
]

👉 The threshold with **maximum Information Gain** is selected for splitting.


## 7. Gini Index (Alternative Split Criterion)

Another measure used in Decision Trees is **Gini Index**.

Gini measures how impure a node is.

Lower Gini value = Better split

Decision Tree chooses the threshold that gives **minimum weighted Gini**.


## 8. Final Decision Tree Split

After calculating Information Gain / Gini for all thresholds:

👉 Select the threshold that gives:

- Maximum Information Gain  
OR  
- Minimum Gini Index  

That threshold becomes the **decision rule**:

Temperature ≤ Best_T  → Left branch  
Temperature > Best_T → Right branch


## 9. Key Points (Exam Ready)

- Numerical features are split using **thresholds**.
- Thresholds are calculated using **midpoints of sorted values**.
- Best threshold is selected using:
  - Information Gain (ID3, C4.5)
  - Gini Index (CART)
- Each split is **binary**.
- This process is repeated recursively to grow the tree.
