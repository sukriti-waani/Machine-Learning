# 🌳 Decision Tree Regressor

**Topic:** Decision Tree for Regression  
**Split Criterion:** Variance Reduction (CART)  
**Goal:** Predict continuous values (e.g., Salary)


## 1. What is Decision Tree Regression?

**Decision Tree Regressor** is used when the target variable is **continuous**.

Examples:
- Predicting Salary
- Predicting House Price
- Predicting Temperature

👉 Unlike classification trees (Entropy/Gini),  
**Regression Trees use Variance Reduction** to decide the best split.


## 2. Example Dataset

Feature: Experience (Exp)  
Target: Salary (in K)

| Exp | Category Gap | Salary (K) |
|-----|---------------|------------|
| 1   | Yes           | 40         |
| 2   | Yes           | 42         |
| 2.5 | No            | 52         |
| 3   | No            | 60         |
| 4   | Yes           | 56         |

We will try different thresholds on **Exp** to find the best split.


## 3. Variance (Impurity Measure)

In regression trees, node impurity is measured using **Variance**.

### Formula:
Variance = (1 / n) * Σ (yᵢ − ȳ)²

Where:
- yᵢ = actual output value  
- ȳ = mean of outputs in the node  
- n = number of samples  

👉 Lower variance = more pure node.


## 4. Variance of Root Node

Given Salaries:
40, 42, 52, 60, 56

Mean (ȳ) = (40 + 42 + 52 + 60 + 56) / 5 = 50

Variance(root)  
= (1/5) * [  
  (40−50)² + (42−50)² + (52−50)² + (60−50)² + (56−50)²  
]

= (1/5) * [100 + 64 + 4 + 100 + 36]  
= 304 / 5  
= **60.8**


## 5. Split 1: Threshold = Exp ≤ 2

Left Child (Exp ≤ 2):  
Salaries = [40]

Right Child (Exp > 2):  
Salaries = [42, 52, 60, 56]

Variance(Left) = 100  
Variance(Right) = (1/4) * [(42−50)² + (52−50)² + (60−50)² + (56−50)²]  
= (1/4) * [64 + 4 + 100 + 36]  
= 204 / 4  
= **51**


## 6. Variance Reduction for Split 1

Variance Reduction =  
Var(root) − [ (n₁/n)·Var(left) + (n₂/n)·Var(right) ]

= 60.8 − [ (1/5)*100 + (4/5)*51 ]  
= 60.8 − 20 - 40.8
= **0**

👉 This split reduces variance by 0.


## 7. Split 2: Threshold = Exp ≤ 2.5

Left Child (Exp ≤ 2.5):  
Salaries = [40, 42]

Right Child (Exp > 2.5):  
Salaries = [52, 60, 56]

Mean(Left) = (40 + 42) / 2 = 41  
Variance(Left) = (1/2) * [(40−41)² + (42−41)²]  
= (1/2) * [1 + 1]  
= **1**

Mean(Right) = (52 + 60 + 56) / 3 = 56  
Variance(Right) = (1/3) * [(52−56)² + (60−56)² + (56−56)²]  
= (1/3) * [16 + 16 + 0]  
= **10.66**


## 8. Variance Reduction for Split 2

Variance Reduction  
= 60.8 − [ (2/5)*1 + (3/5)*10.66 ]

= 60.8 − [0.4 + 6.396]  
= 60.8 − 6.796  
= **54.004 (approx)**

👉 This split gives **much higher variance reduction**.


## 9. Best Split Selection

Compare Variance Reduction:

- Split at Exp ≤ 2   → VR = 20  
- Split at Exp ≤ 2.5 → VR ≈ 54  

👉 Best split is **Exp ≤ 2.5**  
because it gives **maximum variance reduction**.


## 10. Prediction at Leaf Nodes

In Decision Tree Regression,  
the output at each leaf node is the **mean of target values**.

Example:

If leaf contains salaries [40, 42]  
Prediction = (40 + 42) / 2 = **41**

If leaf contains [52, 60, 56]  
Prediction = (52 + 60 + 56) / 3 = **56**


## 11. Exam-Ready Summary

- Decision Tree Regressor is used for **continuous outputs**.
- Node impurity is measured using **Variance**.
- Best split is chosen using **Maximum Variance Reduction**.
- Splits are binary: Feature ≤ Threshold vs Feature > Threshold.
- Final prediction at a leaf node is the **mean of target values**.

👉 Regression Trees use CART algorithm with variance as splitting criterion.
