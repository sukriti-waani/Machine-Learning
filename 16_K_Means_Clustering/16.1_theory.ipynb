# 📘 K-Means Clustering
---

## 1️⃣ What is K-Means Clustering?

K-Means is an **unsupervised machine learning algorithm** used for **clustering**.

- It groups similar data points into **K clusters**.
- Each cluster is represented by a **centroid** (center point).
- Points inside a cluster are **more similar to each other** than to points in other clusters.
- K-Means works best when clusters are roughly **round and well-separated**.

---

## 2️⃣ Where is K-Means Used?

- Customer segmentation (group customers by behavior)
- Image segmentation
- Market basket analysis
- Document clustering
- Grouping similar users or products
- Pattern discovery in unlabeled data

---

## 3️⃣ Intuition of K-Means

- Initially, data points are scattered.
- We choose a value of K (for example, K = 2).
- The algorithm tries to divide the points into **2 natural groups**.
- Points closer to one center form **Cluster 1**.
- Points closer to the other center form **Cluster 2**.
- Over iterations, clusters become tighter and more meaningful.

---

## 4️⃣ Step-by-Step Working of K-Means Algorithm

The K-Means algorithm follows these steps:

### Step 1: Choose the number of clusters (K)
- Decide how many clusters you want to form.
- Example: K = 2, K = 3, etc.

### Step 2: Initialize centroids
- Randomly pick K points as the **initial centroids** (cluster centers).

### Step 3: Assign points to nearest centroid
- For each data point, compute its distance from all centroids.
- Assign the point to the cluster whose centroid is **closest**.

### Step 4: Update centroids
- For each cluster, move the centroid to the **average position** of all points in that cluster.

### Step 5: Repeat
- Repeat Step 3 and Step 4 until:
  - Centroids do not change much, or  
  - Cluster assignments stop changing.

This means the algorithm has **converged**.

---

## 5️⃣ Distance Measures in K-Means

### (a) Euclidean Distance
- Straight-line distance between two points.
- Used when features are continuous and scaled properly.
- Example use case: Air travel (straight distance between cities).

### (b) Manhattan Distance
- Distance measured along axes (like moving in a grid).
- Used when movement is restricted to horizontal and vertical paths.
- Example use case: City blocks or street navigation.

In K-Means, **Euclidean distance** is most commonly used.

---

## 6️⃣ Choosing the Best Value of K (Elbow Method)


To select the best value of K, we use **WCSS (Within-Cluster Sum of Squares)**.

- WCSS measures how compact the clusters are.
- For different values of K (like 1 to 10), we compute WCSS.
- As K increases, WCSS decreases.
- We plot K vs WCSS and look for a point where the decrease becomes slow.
- This point looks like an **“elbow”** in the graph.

👉 The value of K at this elbow point is chosen as the **optimal number of clusters**.

---

## 7️⃣ Random Initialization Trap


- K-Means starts with **random centroids**.
- Different random starts can lead to **different final clusters**.
- This can cause:
  - Poor clustering
  - Getting stuck in local minima

This problem is called the **Random Initialization Trap**.

---

## 8️⃣ K-Means++ Initialization

To overcome random initialization issues, **K-Means++** is used.

- It chooses the first centroid randomly.
- The next centroids are chosen such that they are **far away from existing centroids**.
- This spreads out the initial centroids.
- Results in:
  - Faster convergence
  - Better and more stable clustering

---

## 9️⃣ Advantages of K-Means

- Simple and easy to understand
- Fast and efficient for large datasets
- Easy to implement
- Works well when clusters are clearly separated

---

## 🔟 Disadvantages of K-Means

- Need to specify K beforehand
- Sensitive to initial centroid selection
- Does not work well for:
  - Non-spherical clusters
  - Clusters of different sizes
  - Overlapping clusters
- Sensitive to outliers
- Results can change with different initializations

---

## 1️⃣1️⃣ When to Use K-Means?

Use K-Means when:

- Data is unlabeled
- Clusters are expected to be roughly round
- Dataset is large
- You need a fast clustering method
- Features are numeric and scaled
