# 📊 Feature Selection vs Feature Extraction — Machine Learning

## 📌 What is Feature Selection?
**Feature Selection** is the process of choosing a **subset of the original features** that are most relevant for building a machine learning model.  
It keeps only those features that contribute most to prediction and removes the rest.

### 🔹 Key Points
- Selects a subset of the **original features**  
- Helps reduce overfitting and training time  
- Improves model interpretability (because features remain unchanged)  
- Common methods: Filter, Wrapper, Embedded techniques

---

## 📌 What is Feature Extraction?
**Feature Extraction** is the process of **transforming original features into a new set of features**.  
The new features capture important information while reducing dimensionality.

### 🔹 Key Points
- Transforms original features into a **new feature space**  
- New features may not be directly interpretable  
- Useful when raw data is high-dimensional or complex  
- Common techniques: PCA, LDA, Autoencoders

---

## 🔍 Feature Selection vs Feature Extraction — Comparison

| Aspect | Feature Selection | Feature Extraction |
|--------|-------------------|--------------------|
| What it does | Keeps a subset of original features | Creates new features from original ones |
| Data transformation | No | Yes |
| Goal | Reduce dimensionality by elimination | Reduce dimensionality by transformation |
| Interpretability | Higher (original features) | Lower (new features) |
| Examples | Filter, Wrapper, Embedded | PCA, LDA, Autoencoders |

---

## 🧠 When to Use Each

### ✔ Use Feature Selection When
- You want to **keep original features** for interpretation  
- The dataset isn’t extremely high-dimensional  
- You want to remove irrelevant or redundant features

### ✔ Use Feature Extraction When
- The data is **very high-dimensional or complex**  
- You want to transform features to capture meaningful structure  
- You want stronger dimensionality reduction

---

## 📌 Summary

> **Feature Selection** chooses the most useful features from the original dataset without changing them.  
> **Feature Extraction** transforms the original features into a new reduced set of features.  
> Both methods help reduce dimensionality and improve machine learning performance.
