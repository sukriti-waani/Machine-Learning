# 📘 Principal Component Analysis (PCA) – Complete Notes (Simple Words with Formulas)

---

## 1. What is PCA?

Principal Component Analysis (PCA) is a method used to reduce the number of features in a dataset while keeping most of the useful information.

- PCA converts high-dimensional data into low-dimensional data.
- PCA is an unsupervised learning technique (no output labels are used).
- PCA is a feature extraction method because it creates new features.
- PCA helps in faster training, less overfitting, and better visualization.

---

## 2. Curse of Dimensionality

When the number of features increases:

- Data becomes sparse  
- Models become complex  
- Overfitting increases  
- Accuracy may decrease  

This problem is called the curse of dimensionality.

Example:  
In a housing dataset, adding too many features such as house size, number of rooms, number of bathrooms, nearby shops, schools, etc., may reduce model performance instead of improving it.

---

## 3. Why Dimensionality Reduction is Needed

Dimensionality reduction is done to:

- Reduce the curse of dimensionality  
- Improve model accuracy  
- Reduce training time  
- Reduce overfitting  
- Help in visualization (3D to 2D, 2D to 1D)

---

## 4. Two Ways to Reduce Dimensions

### (a) Feature Selection

- Select only important features.  
- Remove irrelevant or weak features.  
- Original features remain unchanged.  

Example:  
Removing features that have almost no relationship with the output variable.

---

### (b) Feature Extraction (PCA)

- Create new features from existing features.  
- Combine multiple features into fewer new features.  
- PCA is a feature extraction technique.  

Example:  
Combining room size and number of rooms to form a new feature like house size.

---

## 5. Covariance and Correlation

### Covariance

Covariance shows how two features vary together.

Covariance of X and Y is calculated as:  
Sum of (each X value minus mean of X) multiplied by (each Y value minus mean of Y), divided by (N minus 1)

Interpretation:
- Positive covariance: both increase together  
- Negative covariance: one increases, other decreases  
- Near zero: no relationship  

---

### Correlation

Correlation shows the strength of the relationship.

Correlation of X and Y equals covariance of X and Y divided by (standard deviation of X multiplied by standard deviation of Y)

Range of correlation: from -1 to +1

- +1 means strong positive relation  
- -1 means strong negative relation  
- 0 means no relation  

---

## 6. PCA Geometric Intuition

Imagine data points in a 2D space.

PCA finds a new direction (line) such that:

- Data spread along this direction is maximum  
- This direction captures the maximum information  

This new direction is called the first principal component (PC1).

The second principal component (PC2) is perpendicular to PC1 and captures the second highest variance.

---

## 7. Principal Components

- PC1 captures the maximum variance.  
- PC2 captures the second maximum variance.  
- PC3 captures the third maximum variance.  

The importance of each principal component depends on how much variance it captures.

---

## 8. Projection in PCA

Each data point is projected onto the principal component direction.

Projection value of point x onto direction u is calculated as:  
Dot product of x and u divided by the magnitude of u

If u is a unit vector, projection becomes simply the dot product of x and u.

These projected values are used to form new reduced features.

---

## 9. Covariance Matrix

The covariance matrix is a table that shows:

- Variance of each feature on the diagonal  
- Covariance between pairs of features in off-diagonal positions  

For two features X and Y, covariance matrix looks like:

[ Variance of X     Covariance of X and Y  
  Covariance of Y and X   Variance of Y ]

For more features, the matrix becomes larger.

---

## 10. Eigenvalues and Eigenvectors

Eigenvectors are special directions.

Eigenvalues tell how much variance is captured in those directions.

They follow the relationship:

Matrix A multiplied by vector v equals lambda multiplied by vector v

Where:
- A is the covariance matrix  
- v is the eigenvector  
- lambda is the eigenvalue  

In PCA:
- Eigenvector with largest eigenvalue gives PC1  
- Second largest eigenvalue gives PC2  
- And so on  

---

## 11. PCA Algorithm (Step-by-Step in Words)

1. Standardize the dataset so that all features have similar scale.  
2. Compute the covariance matrix of the dataset.  
3. Find eigenvalues and eigenvectors of the covariance matrix.  
4. Sort eigenvalues in descending order.  
5. Select top k eigenvectors (principal components).  
6. Project the original data onto these k principal components to get reduced data.

---

## 12. Dimensionality Reduction Examples

- Converting 3D data into 2D  
- Converting 3D data into 1D  
- Converting 2D data into 1D  

This is done by keeping only the most important principal components.

---

## 13. Final Summary

- PCA reduces dimensions while keeping important information.  
- PCA creates new transformed features.  
- PCA is based on variance maximization.  
- PCA uses covariance, eigenvalues, and eigenvectors.  
- PCA improves speed, accuracy, and visualization.  

---

✅ In short: PCA finds new directions that capture maximum data spread and projects data onto those directions to reduce dimensions.
