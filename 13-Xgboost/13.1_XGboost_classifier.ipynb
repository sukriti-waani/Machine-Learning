# XGBoost Classifier (Extreme Gradient Boosting)

## What is XGBoost?
XGBoost (Extreme Gradient Boosting) is a powerful ensemble machine learning algorithm used for:
- Classification
- Regression
- Ranking problems

It works by combining multiple weak learners (decision trees) in a sequential manner to build a strong predictive model.

## Key Idea
Each new tree is trained to correct the mistakes (residuals) made by the previous trees.

Final prediction = Sum of predictions from all trees + base model output

# Workflow of XGBoost (Classification)

## Steps Involved

1. Construct a Base Model  
   - Initial prediction is usually a constant value.
   - For classification, base prediction = 0.5 (probability)

2. Calculate Residuals  
   Residual = Actual Output - Predicted Output

3. Build a Decision Tree on Residuals  
   - Tree splits data using features like Salary, Credit, etc.
   - Leaves store residual values

4. Calculate Similarity Weights  
   Used to decide the best splits

5. Compute Gain  
   Gain determines whether a split improves the model

6. Update Final Prediction  
   New Output = Base Prediction + Learning Rate × Tree Output


# Residuals in XGBoost

Residual = Actual Value (y) - Predicted Value (ŷ)

In classification:
- Initial predicted probability (ŷ) = 0.5
- Residuals become:
  - If y = 1 → residual = 1 - 0.5 = +0.5
  - If y = 0 → residual = 0 - 0.5 = -0.5

Residuals help the next decision tree learn:
"What mistakes did the previous model make?"


# Similarity Score (Leaf Weight)

## Formula:
Similarity = (Σ Residuals)² / Σ [p × (1 - p)] + λ

Where:
- p = predicted probability
- λ = regularization term (hyperparameter)
- Residuals = (Actual - Predicted)

## Purpose:
- Measures how good a leaf node is
- Higher similarity = better grouping of similar residuals


# Gain in XGBoost

## Formula:
Gain = Similarity(Left Node) + Similarity(Right Node) - Similarity(Parent Node)

## Purpose:
- Used to decide the best split in a decision tree
- Higher gain means better split

If Gain > 0:
- Split is accepted

If Gain ≤ 0:
- Split is rejected


# Logistic (Sigmoid) Function

In XGBoost Classification, output is passed through sigmoid:

σ(z) = 1 / (1 + e^(-z))

Where:
z = Base prediction + learning rate × tree output

## Purpose:
- Converts raw scores into probabilities
- Output range: (0, 1)

Example:
If z = 0.1  
σ(0.1) ≈ 0.52


# Final Prediction in XGBoost Classifier

Final Output = σ(Base Score + η × DT1 + η × DT2 + η × DT3 + ...)

Where:
- σ = Sigmoid function
- η = Learning Rate
- DT = Decision Trees

Each new tree improves the prediction gradually.


# Advantages of XGBoost

- High Accuracy
- Handles missing values
- Prevents overfitting using regularization
- Fast training using parallel processing
- Works well on structured/tabular data
- Supports both classification and regression


# Important Hyperparameters in XGBoost

- learning_rate (η): Controls how much each tree contributes
- max_depth: Depth of trees
- n_estimators: Number of trees
- lambda (λ): Regularization term
- subsample: Fraction of data used per tree
- colsample_bytree: Fraction of features used per tree


# Summary

XGBoost is an ensemble boosting algorithm that builds decision trees sequentially.
Each new tree learns from the residuals (errors) of previous trees.
Gain is used to select the best splits.
Final output is passed through a sigmoid function to convert it into probability.
Regularization and learning rate control overfitting.
