# XGBoost Regressor (Extreme Gradient Boosting for Regression)

## What is XGBoost Regressor?
XGBoost Regressor is an ensemble learning algorithm used for predicting continuous values.
It builds multiple decision trees sequentially, where each new tree learns from the errors
(residuals) of the previous trees.

## Key Idea
Final Prediction = Base Prediction + η × (DT1 + DT2 + DT3 + ...)

Where:
- η = learning rate
- DT = decision trees


# Dataset and Base Model in XGBoost Regressor

Example features (from PDF):
- Experience (Exp)
- Gap (Yes/No)
- Salary (Target Variable)

## Step 1: Create a Base Model
Base model prediction = Mean of target values (Salary)

Example:
If average salary = 51K  
Base prediction for all data points = 51K


# Residual Computation

Residual = Actual Value - Predicted Value

Example:
Actual Salary = 40K  
Base Prediction = 51K  

Residual = 40 - 51 = -11

Residuals represent how much the model is wrong.
These residuals become the target for the next decision tree.


# Decision Tree Construction on Residuals

## Step 2: Construct DT1 using features like:
- Experience (Exp)
- Gap

The tree splits the data to group similar residuals together.

Example splits:
- Exp ≤ 2.5
- Exp > 2.5

Each leaf stores the average of residuals in that region.

# Similarity Weight (Leaf Weight)

## Formula:
Similarity Weight = (Σ Residuals)² / (Number of Residuals + λ)

Where:
- λ = regularization parameter (hyperparameter)

Purpose:
- Measures how good a leaf node is
- Higher similarity means better grouping of residuals


# Gain Calculation

## Formula:
Gain = Similarity(Left Child) + Similarity(Right Child) - Similarity(Parent)

## Purpose:
- Helps select the best feature and split point
- Split is accepted if Gain is high


# Learning Rate (η)

Learning Rate controls how much each tree contributes to the final prediction.

Example:
If DT output = 5  
η = 0.1  

Contribution = 0.1 × 5 = 0.5

Smaller η:
- Slower learning
- Better generalization


# Final Output of XGBoost Regressor

Final Output = Base Prediction + η × DT1 + η × DT2 + ... + η × DTn

Example:
Base Prediction = 51  
DT1 Output = 5  
η = 0.1  

Final Output = 51 + 0.1 × 5 = 51.5


# Regularization in XGBoost

Regularization reduces overfitting.

λ (lambda) is added in denominator of similarity weight:

Similarity = (Σ Residuals)² / (No. of Residuals + λ)

Higher λ:
- Penalizes complex trees
- Makes model simpler and more robust


# XGBoost Regressor – Step-by-Step Flow

1. Build base model (mean prediction)
2. Compute residuals
3. Train decision tree on residuals
4. Calculate similarity and gain
5. Add tree output to base model using learning rate
6. Repeat steps to build multiple trees
7. Final output = sum of all tree contributions


# Summary

XGBoost Regressor predicts continuous values using boosting.
Each new tree learns from residuals of previous trees.
Gain decides best splits.
Learning rate controls contribution of each tree.
Regularization reduces overfitting.
Final prediction is the sum of base prediction and tree outputs.
