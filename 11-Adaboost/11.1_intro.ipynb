# AdaBoost (Adaptive Boosting)

## 1. What is Ensemble Learning?
Ensemble learning means combining multiple simple models (called weak learners) to build a stronger and more accurate model (called a strong learner).

Two common ensemble techniques:
- **Bagging** (e.g., Random Forest): Models are trained independently and combined by voting/averaging.
- **Boosting** (e.g., AdaBoost, Gradient Boosting, XGBoost): Models are trained **sequentially**, where each new model focuses on correcting the mistakes of the previous ones.

---

## 2. Decision Trees and Their Problems
A full decision tree:
- Has **low bias** (can fit complex patterns)
- Has **high variance** (easily overfits)

A very small tree (Decision Tree Stump: depth = 1):
- Has **high bias** (too simple, underfits)
- Has **low variance**

AdaBoost uses **decision tree stumps** as weak learners and combines many of them to reduce bias while controlling variance.

---

## 3. What is a Weak Learner?
A weak learner is a model that performs only slightly better than random guessing.

In AdaBoost:
- Weak learners are usually **decision tree stumps** (trees with only one split).
- Individually, they are weak.
- Together, they form a **strong classifier**.

---

## 4. Basic Idea of AdaBoost
AdaBoost works by:
1. Training a weak learner.
2. Finding which data points are misclassified.
3. Increasing the importance (weight) of misclassified points.
4. Training the next learner to focus more on these difficult points.
5. Repeating this process multiple times.
6. Combining all weak learners using weighted voting.

Final prediction is a weighted sum of all weak learners.

---

## 5. Mathematical Form of AdaBoost (Intuition)

The final model is:

f(x) = α₁h₁(x) + α₂h₂(x) + α₃h₃(x) + ... + αₙhₙ(x)

Where:
- h₁, h₂, ..., hₙ are weak learners (stumps)
- α₁, α₂, ..., αₙ are weights showing how important each weak learner is
- Learners with lower error get higher weight

---

## 6. Step-by-Step AdaBoost Algorithm

### Step 1: Initialize Sample Weights
Initially, all data points are given equal weights.

If there are N samples:
Weight of each sample = 1/N

---

### Step 2: Train First Decision Stump
Train a simple decision tree stump using weighted data.
Choose the split using entropy or Gini index.

---

### Step 3: Compute Error of the Stump
Total Error (TE) is the sum of weights of misclassified samples.

TE = sum of weights of wrongly predicted points

---

### Step 4: Compute Performance (Alpha) of the Stump

α = 1/2 * ln((1 − TE) / TE)

If error is low, α is high.
This means the model is more important in final prediction.

---

### Step 5: Update Sample Weights
- Increase weights of misclassified points.
- Decrease weights of correctly classified points.

This forces the next model to focus more on difficult samples.

---

### Step 6: Normalize Weights
After updating, weights are normalized so that their sum becomes 1.

---

### Step 7: Repeat
Repeat the process for multiple stumps.

---

## 7. Final Prediction

For a new input:
Each stump gives a prediction (Yes/No or +1/−1).
Final output is decided by **weighted voting**:

If sum of weights for "Yes" > sum of weights for "No"  
Then output = Yes  
Else output = No

---

## 8. Bias–Variance Behavior

Decision stump alone:
- High bias
- Low variance

AdaBoost (many stumps combined):
- Lower bias
- Controlled variance
- Better generalization

---

## 9. Advantages of AdaBoost
- Improves accuracy by focusing on hard examples.
- Works well with simple models.
- Often performs better than a single decision tree.
- Less prone to overfitting than deep trees in many cases.

---

## 10. Disadvantages of AdaBoost
- Sensitive to noisy data and outliers.
- If data has wrong labels, AdaBoost may focus too much on them.
- Training is sequential, so it is slower than bagging.

---

## 11. Comparison: Bagging vs Boosting

| Aspect        | Bagging (Random Forest) | Boosting (AdaBoost) |
|--------------|--------------------------|---------------------|
| Training     | Parallel                 | Sequential          |
| Focus        | Reduce variance          | Reduce bias         |
| Data weights | Same for all models      | Updated every round |
| Overfitting  | Good control             | Sensitive to noise  |

---

## 12. Key Exam Points

- AdaBoost is a **boosting algorithm** that combines weak learners into a strong learner.
- It assigns **higher weight to misclassified samples**.
- Final prediction is a **weighted sum of weak learners**.
- Decision stumps are commonly used as base learners.
- AdaBoost reduces bias and improves accuracy but is sensitive to noisy data.
