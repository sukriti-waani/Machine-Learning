```python
from sklearn.cluster import DBSCAN 
from sklearn.datasets import make_moons 
import matplotlib.pyplot as plt 
%matplotlib inline 
```

```python
# Generate a synthetic 2D dataset with two interleaving half-moon shapes
# n_samples=250 -> total number of data points to generate
# noise=0.05    -> adds small random noise to make the data more realistic
# X             -> feature matrix (coordinates of the points)
# y             -> true labels (useful only for visualization/evaluation, not for unsupervised learning)

X, y = make_moons(n_samples=250, noise=0.05)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y)
```

```python
# feature scaling 
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
# Fit the scaler on the feature data X and transform it
# This scales each feature to a similar range (e.g., mean = 0, std = 1 for StandardScaler)
# Scaling is important for distance-based algorithms like K-Means, DBSCAN, and Hierarchical Clustering

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
# Create a DBSCAN clustering model
# eps=0.3 -> defines the neighborhood radius (maximum distance to consider points as neighbors)
# Points within this distance are grouped into the same dense region (cluster)

dbscan = DBSCAN(eps=0.3)
```

```python
# Fit the DBSCAN model on the scaled feature data
# This step lets DBSCAN identify dense regions (clusters) and noise (outliers)
# DBSCAN does not require the number of clusters in advance

dbscan.fit(X_scaled)
```

```python
# dbscan.labels_ contains the cluster label assigned to each data point
# Each element corresponds to one point in X_scaled
# Labels: 0, 1, 2, ... -> different clusters
# Label: -1 -> noise/outlier points that do not belong to any cluster

dbscan.labels_
```

```python
plt.scatter(X[:, 0], X[:, 1], c=dbscan.labels_)
```
