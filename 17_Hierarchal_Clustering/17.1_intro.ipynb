# 🌳 Hierarchical Clustering

## 📌 What is Hierarchical Clustering?
Hierarchical Clustering is an **unsupervised learning** technique that groups data points into clusters by building a **tree-like structure** called a **dendrogram**.  
Unlike K-Means, it **does not require choosing K (number of clusters) in advance**. We can decide the number of clusters later by cutting the dendrogram at a chosen height (distance threshold).

**Key idea:**  
- Build clusters step-by-step.
- Visualize how points/clusters merge or split.
- Choose clusters by cutting the dendrogram.

---

## 🧭 Types of Hierarchical Clustering

### 1️⃣ Agglomerative (Bottom-Up)
- Start with **each data point as its own cluster**.
- Repeatedly **merge the two closest clusters**.
- Continue until **all points form one big cluster**.
- This is the **most commonly used** hierarchical method.

**Steps (as shown in the diagram):**
1. Treat each point as a separate cluster.  
2. Find the **nearest pair of clusters** using a distance metric.  
3. Merge them into a new cluster.  
4. Repeat until only one cluster remains.  

---

### 2️⃣ Divisive (Top-Down)
- Start with **all data points in one single cluster**.
- Repeatedly **split the cluster** into smaller clusters.
- Continue splitting until each point becomes its own cluster (or until desired clusters are formed).

**Note:**  
Divisive methods are **less common** and computationally more expensive than agglomerative methods.

---

## 📏 Distance Metrics (How “closeness” is measured)
To decide which clusters to merge/split, we use distance measures:
- **Euclidean Distance** (most common)
- Manhattan Distance
- Cosine Distance (for text/embeddings)

**Why it matters:**  
The distance metric affects the final cluster structure.

---

## 🔗 Linkage Criteria (How distance between clusters is computed)
When clusters contain multiple points, distance between clusters can be defined in different ways:

- **Single Linkage:**  
  Distance between the **closest pair** of points from two clusters  
  → Can create long “chains”.

- **Complete Linkage:**  
  Distance between the **farthest pair** of points  
  → Produces compact clusters.

- **Average Linkage:**  
  Average distance between all pairs across two clusters.

- **Ward’s Method:**  
  Merges clusters that result in the **minimum increase in total variance**  
  → Often produces balanced clusters.

---

## 🌲 Dendrogram (Tree Diagram)
A **dendrogram** visually shows:
- How clusters are merged (or split)
- The **distance (height)** at which merges happen

**How to choose the number of clusters:**
- Draw a **horizontal cut** across the dendrogram.
- The number of vertical branches cut = **number of clusters**.
- A common rule: choose the **largest vertical gap** (longest vertical line without a merge crossing it) and cut there.

**Threshold concept:**  
- The cut height acts as a **distance threshold**.  
- Points/clusters merged above this threshold are considered separate clusters.

---

## 🧠 Important Characteristics
- ❌ **No centroids** (unlike K-Means).  
- ✅ Works well when cluster shapes are **not spherical**.  
- ✅ Produces a **hierarchy of clusters** (useful for exploration).  
- ❌ **Computationally expensive** for large datasets (time & memory).  
- ❌ Sensitive to **noise and outliers** (outliers can form their own branches).

---

## ⚖️ Advantages
- No need to predefine **K**.  
- Easy to **visualize** with dendrograms.  
- Good for **exploratory data analysis**.  
- Can reveal **nested cluster structures**.

---

## ⚠️ Limitations
- Not scalable for very large datasets (O(n²) or worse).  
- Once a merge is done, it **cannot be undone** (greedy nature).  
- Results depend heavily on **distance metric and linkage method**.

---

## 🛠️ Practical Workflow (Python – Conceptual)
1. Scale features (important for distance-based methods).
2. Compute distance matrix.
3. Apply **Agglomerative Clustering** (choose linkage).
4. Plot **dendrogram**.
5. Cut dendrogram to choose clusters.
6. Analyze clusters.

---

## 📊 Use Cases
- Customer segmentation  
- Document clustering  
- Gene expression analysis  
- Market basket analysis  
- Image grouping

---

## 📝 Short Summary
- Hierarchical Clustering builds a **tree of clusters (dendrogram)**.  
- Two types: **Agglomerative (bottom-up)** and **Divisive (top-down)**.  
- Distance metrics + linkage methods decide how clusters form.  
- Number of clusters is chosen by **cutting the dendrogram at a threshold**.  
- No centroids, no pre-defined K, but **computationally costly** for big data.
