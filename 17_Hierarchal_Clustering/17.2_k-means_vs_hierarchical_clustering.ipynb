## 🔁 K-Means vs Hierarchical Clustering

| Feature | K-Means Clustering | Hierarchical Clustering |
|--------|-------------------|--------------------------|
| Type | Partition-based clustering | Hierarchy-based clustering |
| Learning | Unsupervised | Unsupervised |
| Need to specify K | Yes, K must be given before training | No, K is chosen later using dendrogram |
| Cluster structure | Flat clusters | Tree-like structure (dendrogram) |
| Centroids | Uses centroids (means of clusters) | No centroids |
| Distance metric | Usually Euclidean distance | Can use various distance metrics |
| Speed & Scalability | Fast and scalable for large datasets | Slow and memory-heavy for large datasets |
| Cluster shape | Works best for spherical clusters | Can handle complex cluster shapes |
| Sensitivity to initialization | Sensitive to initial centroids | No random initialization |
| Outliers | Sensitive to outliers | Outliers can form separate branches |
| Visualization | No natural visual hierarchy | Dendrogram provides clear visualization |
| When to use | Large datasets with clear, compact clusters | Small to medium datasets, exploratory analysis |

---

### ✅ K-Means – Key Points
- Efficient and fast for **large datasets**.  
- Requires choosing **K in advance** (Elbow/Silhouette methods help).  
- Works best when clusters are **compact and well-separated**.  
- Sensitive to **initialization and outliers**.

---

### 🌳 Hierarchical Clustering – Key Points
- Builds a **dendrogram** showing nested clusters.  
- **No need to predefine K**; choose clusters by cutting the dendrogram.  
- Useful for **exploratory data analysis** and understanding structure.  
- Computationally **expensive** for large datasets.

---

### 📝 One-line Summary
- **K-Means** is fast and scalable but needs K beforehand.  
- **Hierarchical Clustering** is more interpretable with dendrograms but slower for large data.
