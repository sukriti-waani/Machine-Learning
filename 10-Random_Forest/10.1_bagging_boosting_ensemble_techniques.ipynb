# 🌲 Random Forest: Classification, Regression & Out-of-Bag (OOB) Evaluation

**Topics Covered:**
- Random Forest (Bagging + Feature Sampling)
- Random Forest for Classification & Regression
- Bias–Variance Reduction
- Out-of-Bag (OOB) Evaluation


## 1. What is Random Forest?

**Random Forest** is an ensemble learning method that builds **many decision trees** and combines their predictions.

Each tree is trained on:
- A **random sample of rows (data points)** using sampling with replacement (Bagging)
- A **random subset of features** at each split (Feature Sampling)

👉 Final prediction is obtained by **combining outputs of all trees**.


## 2. Bagging + Feature Sampling

Two sources of randomness are used:

1. **Row Sampling (Bagging):**
   - Randomly sample data points **with replacement**
   - Each tree gets a different training set

2. **Feature Sampling:**
   - At each node, only a **subset of features** is considered for splitting

👉 This makes trees **less correlated** and improves overall performance.


## 3. Random Forest for Classification

In **classification**, each tree predicts a class label.

**Final Output = Majority Voting**

Example:
- Tree 1 → Class A  
- Tree 2 → Class B  
- Tree 3 → Class A  

Final Prediction → **Class A (majority vote)**

👉 This reduces the chance of wrong prediction by any single tree.


## 4. Random Forest for Regression

In **regression**, each tree predicts a numerical value.

**Final Output = Average of all tree predictions**

Example:
- Tree 1 → 50  
- Tree 2 → 52  
- Tree 3 → 48  

Final Prediction = (50 + 52 + 48) / 3 = **50**

👉 Averaging reduces variance and improves prediction stability.


## 5. Why Use Random Forest Instead of a Single Decision Tree?

### 🔴 Decision Tree Problems:
- Tends to **overfit**
- High variance (very sensitive to data changes)
- Good training accuracy but poor test accuracy

### ✅ Random Forest Advantages:
- Combines many trees → **generalized model**
- **Low variance**
- Better test accuracy
- More robust to noise

👉 Random Forest reduces overfitting of individual decision trees.


## 6. Bias–Variance Tradeoff

- Decision Tree:
  - Low bias
  - High variance (overfitting)

- Random Forest:
  - Low bias
  - **Low variance** (due to averaging/voting)

👉 Ensemble of trees smooths out errors of individual trees.


## 7. Out-of-Bag (OOB) Evaluation

During bagging:
- Each tree is trained on a bootstrap sample (~63% of data)
- The remaining ~**37% of data is not used** for that tree
- This unused data is called **Out-of-Bag (OOB) data**

👉 OOB data acts like a **validation set** for that tree.


## 8. How OOB Score is Computed

Steps:
1. For each data point, collect predictions from trees **where this point was OOB**.
2. Combine those predictions:
   - Classification → Majority vote
   - Regression → Average
3. Compare predicted vs actual values.
4. Compute accuracy (classification) or error (regression).

👉 This gives **OOB Score**, an unbiased estimate of model performance.


## 9. OOB vs Train-Test Split

**Train-Test Split:**
- Manually split data into train and test sets
- Reduces training data size

**OOB Evaluation:**
- Uses data already left out during bagging
- No need for separate validation set
- Efficient use of data

👉 OOB is especially useful when dataset is small.


## 10. OOB in scikit-learn

In scikit-learn, enable OOB using:

oob_score = True

Example:
RandomForestClassifier(oob_score=True)

Then access:
model.oob_score_

👉 This gives OOB accuracy for classification
and OOB R² score for regression.


## 11. Workflow of Random Forest with OOB

1. Given dataset D
2. Create many bootstrap samples (bagging)
3. Train one decision tree on each sample
4. Use OOB data to evaluate each tree
5. Combine tree predictions:
   - Classification → Majority voting
   - Regression → Averaging
6. Use OOB score to estimate generalization performance


## 12. Summary

- Random Forest is an ensemble of Decision Trees.
- Uses:
  - Bagging (row sampling with replacement)
  - Feature sampling at each split
- Classification → Majority voting
- Regression → Average of predictions
- Reduces overfitting and variance.
- OOB evaluation uses left-out samples during bagging as validation data.
- OOB score provides an unbiased estimate of model performance without separate validation set.
