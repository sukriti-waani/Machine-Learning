```python
# Getting all different types of features
# This section separates columns into numerical, categorical, discrete, and continuous types

# Select all numerical columns by excluding object (text) type
num_features = [feature for feature in df.columns if df[feature].dtype != 'O']
print("Num of Numerical Features :", len(num_features))

# Select all categorical (text/object) columns
cat_features = [feature for feature in df.columns if df[feature].dtype == 'O']
print("Num of Categorical Features :", len(cat_features))

# From numerical features, select discrete features
# Discrete features have limited number of unique values (here ≤ 25)
discrete_features = [feature for feature in num_features if len(df[feature].unique()) <= 25]
print("Num of Discrete Features :", len(discrete_features))

# From numerical features, select continuous features
# Continuous features are those numerical features that are not discrete
continuous_features = [feature for feature in num_features if feature not in discrete_features]
print("Num of Continuous Features :", len(continuous_features))


# Numerical features are number-type columns.
# Categorical features are text-type columns.
# Discrete features are numerical columns with few unique values.
# Continuous features are numerical columns with many unique values.
```

## Feature Encoding and Scaling
**One Hot Encoding for Columns which had lesser unique values and not ordinal**
* One hot encoding is a process by which categorical variables are converted into a form that could be provided to ML algorithms to do a better job in prediction.

```python
len(df['model'].unique())
```

```python
df['model'].value_counts()
```

```python
# Import LabelEncoder to convert categorical labels into numeric form
from sklearn.preprocessing import LabelEncoder

# Create a LabelEncoder object
le = LabelEncoder()

# Convert the categorical values in the 'model' column into numeric labels
# Each unique category is assigned a unique number
X['model'] = le.fit_transform(X['model'])
```

```python
X.head()
```

```python
len(df['seller_type'].unique()), len(df['fuel_type'].unique()), len(df['transmission_type'].unique())
```

```python
# Select all numerical columns from X for scaling
num_features = X.select_dtypes(exclude="object").columns

# Manually specify categorical columns that need One-Hot Encoding
onehot_columns = ['seller_type', 'fuel_type', 'transmission_type']

# Import preprocessing tools for encoding and scaling
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

# Create a scaler to standardize numerical features
numeric_transformer = StandardScaler()

# Create One-Hot Encoder for selected categorical features
# drop='first' is used to avoid redundant dummy columns
oh_transformer = OneHotEncoder(drop='first')

# Combine preprocessing steps for different column types
# OneHotEncoder is applied to selected categorical columns
# StandardScaler is applied to numerical columns
# remainder='passthrough' keeps other columns unchanged
preprocessor = ColumnTransformer(
    [
        ("OneHotEncoder", oh_transformer, onehot_columns),
        ("StandardScaler", numeric_transformer, num_features)
    ],
    remainder='passthrough'
)
```

```python
# Fit the preprocessor on X and transform X using the learned scaling and encoding rules
X = preprocessor.fit_transform(X)
```
